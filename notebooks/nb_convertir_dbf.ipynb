```python
import os
from dbfread import DBF
import pandas as pd
```

```python
# Folder containing your DBF files
input_folder = 'C:/SCINCE 2020/SCINCE 2020/15_MEX/tablas'
output_folder = 'C:/SCINCE 2020/SCINCE 2020/15_MEX/tablas_csv'

# Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)
```

```python
# Process all DBF files
all_files = os.listdir(input_folder)
dbf_files = [f for f in all_files if f.lower().endswith('.dbf')]

print(f"Archivos en la carpeta: {len(all_files)}")
print(f"Archivos DBF encontrados: {len(dbf_files)}")

if not dbf_files:
    print("⚠ No se encontraron archivos .dbf en:", input_folder)
else:
    for filename in dbf_files:
        dbf_path = os.path.join(input_folder, filename)
        
        # Read DBF
        dbf = DBF(dbf_path, encoding='latin1')
        df = pd.DataFrame(iter(dbf))
        
        # Create output filename (change extension)
        base_name = os.path.splitext(filename)[0]
        
        # Save as CSV
        csv_path = os.path.join(output_folder, f'{base_name}.csv')
        df.to_csv(csv_path, index=False)
        
        print(f'Converted: {filename} -> {base_name}.csv ({len(df)} filas)')

print("Proceso terminado.")
```
