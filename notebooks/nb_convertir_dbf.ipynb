```python
import os
from dbfread import DBF
import pandas as pd
```

```python
# Folder containing your DBF files
input_folder = 'C:/SCINCE 2020/09_CDMX/tablas'
output_folder = 'C:/SCINCE 2020/09_CDMX/tablas_csv'

# Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)
```

```python
# Process all DBF files
for filename in os.listdir(input_folder):
    if filename.endswith('.dbf') or filename.endswith('.DBF'):
        dbf_path = os.path.join(input_folder, filename)
        
        # Read DBF
        dbf = DBF(dbf_path, encoding='latin1')
        df = pd.DataFrame(iter(dbf))
        
        # Create output filename (change extension)
        base_name = os.path.splitext(filename)[0]
        
        # Save as CSV
        csv_path = os.path.join(output_folder, f'{base_name}.csv')
        df.to_csv(csv_path, index=False)
        
        print(f'Converted: {filename} -> {base_name}.csv')
```
