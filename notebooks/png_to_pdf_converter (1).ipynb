# PNG to PDF Converter

This notebook demonstrates how to convert PNG images to PDF format using Python. You can convert a single image or multiple images into a single PDF.

## Import Libraries

```python
from PIL import Image
import os
from pathlib import Path
from typing import List
```

```python
#!/usr/bin/env python3
"""
Simple PNG to PDF Converter - Quick version

Just import and use: from convert_pngs_simple import convert_pngs_in_folder
"""

from PIL import Image
from pathlib import Path


def convert_pngs_in_folder(folder_path: str):
    """
    Convert all PNG files in a folder to PDF in the same location.
    Simple version with minimal setup.
    
    Usage:
    ------
    convert_pngs_in_folder('./my_images')
    convert_pngs_in_folder('/home/user/pictures')
    """
    
    folder = Path(folder_path)
    png_files = sorted(list(folder.glob('*.png')))
    
    print(f"Converting {len(png_files)} PNG files in {folder_path}...\n")
    
    for png_file in png_files:
        try:
            image = Image.open(png_file)
            
            # Convert RGBA to RGB
            if image.mode in ('RGBA', 'LA', 'P'):
                rgb_image = Image.new('RGB', image.size, (255, 255, 255))
                rgb_image.paste(image, mask=image.split()[-1] if image.mode == 'RGBA' else None)
                image = rgb_image
            
            # Save with same name, .pdf extension
            output_file = png_file.with_suffix('.pdf')
            image.save(output_file, 'PDF')
            print(f"✓ {png_file.name} → {output_file.name}")
        
        except Exception as e:
            print(f"✗ {png_file.name} - Error: {e}")
    
    print(f"\nDone! All PDFs saved in {folder_path}")
```

## Method 1: Convert a Single PNG to PDF

```python
def png_to_pdf_single(input_path: str, output_path: str) -> None:
    """
    Convert a single PNG image to PDF.
    
    Parameters:
    -----------
    input_path : str
        Path to the PNG image file
    output_path : str
        Path for the output PDF file
    
    Returns:
    --------
    None
    
    Example:
    --------
    png_to_pdf_single('image.png', 'output.pdf')
    """
    try:
        # Open the PNG image
        image = Image.open(input_path)
        
        # Convert RGBA to RGB if necessary (PDFs work better with RGB)
        if image.mode in ('RGBA', 'LA', 'P'):
            rgb_image = Image.new('RGB', image.size, (255, 255, 255))
            rgb_image.paste(image, mask=image.split()[-1] if image.mode == 'RGBA' else None)
            image = rgb_image
        
        # Save as PDF
        image.save(output_path, 'PDF')
        print(f"✓ Successfully converted: {input_path} -> {output_path}")
    
    except FileNotFoundError:
        print(f"✗ Error: File not found - {input_path}")
    except Exception as e:
        print(f"✗ Error converting {input_path}: {str(e)}")

# Example usage:
# png_to_pdf_single('my_image.png', 'output.pdf')
```

## Method 2: Convert Multiple PNG Images to a Single PDF

```python
def png_list_to_pdf(input_paths: List[str], output_path: str) -> None:
    """
    Convert multiple PNG images to a single PDF (one image per page).
    
    Parameters:
    -----------
    input_paths : List[str]
        List of paths to PNG image files
    output_path : str
        Path for the output PDF file
    
    Returns:
    --------
    None
    
    Example:
    --------
    png_list_to_pdf(['image1.png', 'image2.png', 'image3.png'], 'combined.pdf')
    """
    try:
        images = []
        
        for path in input_paths:
            try:
                image = Image.open(path)
                
                # Convert RGBA to RGB if necessary
                if image.mode in ('RGBA', 'LA', 'P'):
                    rgb_image = Image.new('RGB', image.size, (255, 255, 255))
                    rgb_image.paste(image, mask=image.split()[-1] if image.mode == 'RGBA' else None)
                    image = rgb_image
                
                images.append(image)
                print(f"✓ Loaded: {path}")
            except Exception as e:
                print(f"✗ Could not load {path}: {str(e)}")
        
        if images:
            # Save all images to a single PDF
            images[0].save(output_path, 'PDF', save_all=True, append_images=images[1:])
            print(f"\n✓ Successfully created PDF: {output_path}")
            print(f"  Total pages: {len(images)}")
        else:
            print("✗ No images were successfully loaded.")
    
    except Exception as e:
        print(f"✗ Error creating PDF: {str(e)}")

# Example usage:
# png_list_to_pdf(['page1.png', 'page2.png', 'page3.png'], 'document.pdf')
```

## Method 3: Convert All PNG Files in a Directory

```python
def convert_directory_pngs_to_pdf(directory: str, output_path: str, sort_files: bool = True) -> None:
    """
    Convert all PNG files in a directory to a single PDF.
    
    Parameters:
    -----------
    directory : str
        Path to directory containing PNG files
    output_path : str
        Path for the output PDF file
    sort_files : bool
        Whether to sort files alphabetically (default: True)
    
    Returns:
    --------
    None
    
    Example:
    --------
    convert_directory_pngs_to_pdf('./images', 'all_images.pdf')
    """
    try:
        # Get all PNG files
        png_files = list(Path(directory).glob('*.png'))
        
        if not png_files:
            print(f"✗ No PNG files found in {directory}")
            return
        
        if sort_files:
            png_files.sort()
        
        print(f"Found {len(png_files)} PNG files.\n")
        
        # Convert to strings and call the list function
        png_paths = [str(f) for f in png_files]
        png_list_to_pdf(png_paths, output_path)
    
    except FileNotFoundError:
        print(f"✗ Error: Directory not found - {directory}")
    except Exception as e:
        print(f"✗ Error: {str(e)}")

# Example usage:
# convert_directory_pngs_to_pdf('./my_images', 'output.pdf')
```

## Additional Features

### Resize Images Before Converting

If your images are very large, you might want to resize them to reduce the PDF file size:

```python
def png_to_pdf_with_resize(input_path: str, output_path: str, max_width: int = 2000, max_height: int = 2000) -> None:
    """
    Convert PNG to PDF with optional resizing.
    
    Parameters:
    -----------
    input_path : str
        Path to the PNG image file
    output_path : str
        Path for the output PDF file
    max_width : int
        Maximum width in pixels (maintains aspect ratio)
    max_height : int
        Maximum height in pixels (maintains aspect ratio)
    """
    try:
        image = Image.open(input_path)
        
        # Resize if necessary
        image.thumbnail((max_width, max_height), Image.Resampling.LANCZOS)
        
        # Convert RGBA to RGB
        if image.mode in ('RGBA', 'LA', 'P'):
            rgb_image = Image.new('RGB', image.size, (255, 255, 255))
            rgb_image.paste(image, mask=image.split()[-1] if image.mode == 'RGBA' else None)
            image = rgb_image
        
        image.save(output_path, 'PDF')
        print(f"✓ Successfully converted and resized: {input_path} -> {output_path}")
    
    except Exception as e:
        print(f"✗ Error: {str(e)}")

# Example usage:
# png_to_pdf_with_resize('large_image.png', 'resized_output.pdf', max_width=1500, max_height=1500)
```

## Notes

- **RGBA to RGB Conversion**: PNG files with transparency (RGBA) are automatically converted to RGB with a white background
- **File Size**: The resulting PDF size depends on image resolution and compression
- **Image Order**: When converting multiple images, they appear in the order provided
- **Compatibility**: The output PDF works with all standard PDF readers
